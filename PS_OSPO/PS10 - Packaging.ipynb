```python
import tomllib

ANSWER = {
    "project": {
        "name": "PS10",
        "version": "1.1.1",
        "dependencies": ["scipy", "matplotlib"],
        "classifiers": [
            "License :: FSF Approved :: GPLv2 License",
            "Private :: Do Not Upload",
        ],
        "optional-dependencies": {
            "docs": ["mkdocs-material"]
        },
    }
}

def verify_subset(actual, expected):
    if isinstance(expected, dict):
        return (
            isinstance(actual, dict)
            and all(
                (key in actual) and verify_subset(actual[key], value)
                for key, value in expected.items()
            )
        )

    if isinstance(expected, list):
        if not isinstance(actual, list):
            return False

        return all(
            any(verify_subset(a, e) for a in actual)
            for e in expected
        )

    return actual == expected # Default

with open("pyproject.toml", "rb") as fopen:
    pyproject = tomllib.load(fopen)

assert verify_subset(pyproject, ANSWER), "Pyproject.toml does not match specifications"
print("Success!")
```
